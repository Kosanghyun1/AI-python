```python
import random

capitals = {
    "한국": "서울",
    "미국": "워싱턴 D.C.",
    "중국": "베이징",
    "일본": "도쿄",
    "영국": "런던",
    "프랑스": "파리",
    "독일": "베를린",
    "브라질": "브라질리아",
    "캐나다": "오타와",
    "호주": "캔버라",
}

# 딕셔너리의 키(국가 이름)들을 리스트로 만듭니다.
nations_list = list(capitals.keys())

# 0부터 (국가 수 - 1) 사이의 무작위 인덱스를 뽑습니다.
random_index = random.randint(0, len(capitals) - 1)

# 선택된 인덱스의 국가와 수도를 가져옵니다.
selected_nation = nations_list[random_index]
selected_capital = capitals[selected_nation]

print(f"선택된 인덱스: {random_index}")
print(f"임의로 선택된 국가({selected_nation})의 수도: {selected_capital}")
```
