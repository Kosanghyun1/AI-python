```python
import matplotlib.pyplot as plt

year = [2014, 2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022]
gdp = [3079.9, 3250.1, 3398.8, 3574, 3678.2, 3721.8, 3744, 4003.6, 4165.5]

plt.figure(figsize=(5, 3))                   # 그래프 크기 (가로 5인치, 세로 3인치)
plt.plot(year, gdp, 'g.-')                   # 초록(g) 점 마커(.) 실선(-)
plt.title('GDP per capita')                  # 그래프 제목
plt.xlabel('years')                          # x축 제목
plt.ylabel('ten thousand won')               # y축 제목
plt.ylim(2500, 4500)                         # y축 범위 (xlim 으로 x축 범위도 가능)
plt.show()
```

```python
import matplotlib.pyplot as plt

height = [156.9, 159.9, 153.5, 151.2, 154.6, 168.3, 165, 154.7, 146, 152.4]
weight = [48.7, 58.5, 48.4, 39, 58.9, 80.8, 59.3, 49.4, 35.3, 51.8]
gender = ['f', 'f', 'f', 'f', 'f', 'm', 'm', 'm', 'm', 'm']

colors = ['r' if a == 'f' else 'b' for a in gender]   # 리스트 내포: f 이면 'r', 아니면 'b'
print(colors)

plt.figure(figsize=(5, 3))
plt.scatter(height, weight, c=colors)        # c 에 색상 리스트를 주면 점마다 색이 달라진다
plt.xlabel('height')
plt.ylabel('weight')
plt.show()
```

```python
import matplotlib.pyplot as plt

temperature = [6, 2.2, 4.6, 7.7, 8, 9.2, 11.7, 12.5, 10.3, 13.5,
               15.2, 5.9, 2.9, 9.2, 8.6, 4.6, 7.2, 7.7, 8.3, 9.5,
               13, 17, 17.4, 12.5, 12.5, 9.4, 7.3, 9.2, 11.8, 13.3, 15]
print('일 수:', len(temperature))

plt.figure(figsize=(8, 3))

plt.subplot(1, 2, 1)
count, edges, _ = plt.hist(temperature)       # 기본 10개 구간. 구간별 도수와 경계값을 돌려준다
plt.title('default (10 bins)')
print('구간별 도수:', [int(c) for c in count])

plt.subplot(1, 2, 2)
plt.hist(temperature, bins=15, alpha=0.5)    # bins: 구간 수, alpha: 투명도
plt.xlabel('temperature')
plt.ylabel('frequency')
plt.title('bins=15, alpha=0.5')

plt.tight_layout()
plt.show()
```

```python

import pandas as pd

def make_df():
    return pd.DataFrame({'name': ['Jessica', 'Liam', 'Sophia', 'Ryan', 'Alex'],
                         'test': [45, 30, 40, 37, 48],
                         'assign1': [20, 17, 22, 18, 24],
                         'assign2': [19, 14, 18, 15, 25]})

df = make_df()

print('-- [코드 11-6] 열 하나 선택 --')
print(df['name'])                            # 한 열을 고르면 시리즈가 된다

print('-- [코드 11-7] 열 여러 개 선택 --')
print(df[['name', 'test']])                  # 열 이름을 리스트로 묶으면 데이터프레임이 된다

print('-- [코드 11-8] pop: 열을 꺼내고 삭제 --')
n = df.pop('name')                           # name 열을 n 에 담고 df 에서는 지운다
print(list(n))
print(list(df.columns))

print('-- [코드 11-9] iterrows: 행마다 반복 --')
df = make_df()                               # name 열이 있는 원래 데이터프레임으로 다시 시작
for index, row in df.iterrows():             # index: 행 인덱스, row: 그 행의 데이터
    print(index, row['name'], row['test'])
```

```python
import numpy as np
import pandas as pd

df = pd.DataFrame({'name': ['Jessica', 'Liam', 'Sophia', 'Ryan', 'Alex'],
                   'gender': ['F', np.nan, np.nan, 'M', 'F'],       # np.nan: 누락된 값
                   'score': [89, 75, 95, np.nan, 93]})
print(df)                                    # [코드 11-22] 누락 값은 NaN 으로 보인다

print(pd.isna(df))                           # [코드 11-23] 누락이면 True
print(pd.isna(df).sum())                     # [코드 11-24] 열마다 누락 개수 (True 는 1 로 계산)

clean = df.dropna()                          # [코드 11-25] 누락이 있는 행을 모두 삭제
print(clean)

clean = df.dropna(subset=['name', 'score'])  # [코드 11-26] name, score 가 누락된 행만 삭제
print(clean)

df.dropna(subset=['name', 'score'], inplace=True)   # [코드 11-27] inplace=True: df 자체에 반영
print(df)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({'nation': ['Korea', 'USA', 'Japan', 'France'],
                   'gold': [12, 36, 4, 6],
                   'silver': [10, 31, 3, 4],
                   'bronze': [11, 27, 7, 6]})
print(df)

df['total'] = df['gold'] + df['silver'] + df['bronze']
print(df[['nation', 'total']])

plt.figure(figsize=(5, 3))
plt.bar(df['nation'], df['gold'], color='gold', label='gold')
plt.bar(df['nation'], df['silver'], color='silver', bottom=df['gold'], label='silver')
plt.bar(df['nation'], df['bronze'], color='darkorange',
        bottom=df['gold'] + df['silver'], label='bronze')
plt.legend()
plt.ylabel('number of medals')
plt.show()
```
